## Problema

## Entrada

else: descuento = 0 total = monto print("\n No se aplica descuento.")

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal | | | |
| Caso límite | | | |
| Caso inválido | | | |


```python
# Escriba la solución aquí
try:
    cliente = int(input)("Ingrese el ID del cliente: ").strip()
    monto = int(input)("Ingrese el monto de la compra: $").strip()
    cliente_frecuente = input("¿ Es cliente frecuente? (S/N): ").strip().upper()
    
    if cliente_frecuente not in ("S", "N"):
        raise ValueError("Respuesta invalida")
    
    if cliente_frecuente.upper() == "S":
        
        descuento = monto + 0.10
        total = monto - descuento
        print(" \Se aplica un descuento del 10%. ")
    else:
        descuento = 0
        total = monto
        print("\nNo se aplica undescuento.")
        print("\n---- FACTURA ----")
        print(f"Cliente: {cliente}")
        print(f"Monto de la compra: ${monto: .2f}")
        print(f"Descuento: ${descuento: .2f}")
        print(f"Total a pagar: ${total:.2f}")
        
except ValueError:
    print("Error: Debe ingresar unicamente valores que se le solicitan en los campos.")
```
